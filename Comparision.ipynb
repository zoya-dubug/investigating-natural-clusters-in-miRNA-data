```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Comparision table
comparison_df = pd.DataFrame({
    "Algorithm": ["K-Means", "GMM", "Hierarchical"],
    "Best_K": [6, 6, 10],
    "Best_Silhouette": [0.259675, 0.263220, 0.260582]
})

comparison_df
```

```python
#Best Silhouette comparision
plt.figure(figsize=(8,5))

sns.barplot(
    data=comparison_df,
    x="Algorithm",
    y="Best_Silhouette"
)

plt.title("Best Silhouette Score by Algorithm")
plt.savefig(
    "screenshots/Best Silhouette Comparision",
    dpi=300,
    bbox_inches="tight"
    )
plt.show()
```

```python
#Best K for each algorithm
k_selection = pd.DataFrame({
    "Algorithm":["K-Means","GMM","Hierarchical"],
    "Preferred_K":[6,6,10]
})

k_selection
```

```python
#Sihouette score table for k = 2 to 10
comparison_scores = pd.DataFrame({
    "K":[2,3,4,5,6,7,8,9,10],
    "KMeans":[0.222,0.228,0.234,0.238,0.260,0.251,0.260,0.257,0.223],
    "GMM":[0.220,0.235,0.245,0.253,0.263,0.254,0.243,0.250,0.251],
    "Hierarchical":[0.208,0.226,0.206,0.202,0.231,0.244,0.258,0.247,0.261]
})
comparison_scores
```

```python
#silhouette vs K plot
plt.figure(figsize=(10,6))

plt.plot(comparison_scores["K"],
         comparison_scores["KMeans"],
         marker="*",
         label="K-Means")

plt.plot(comparison_scores["K"],
         comparison_scores["GMM"],
         marker="*",
         label="GMM")

plt.plot(comparison_scores["K"],
         comparison_scores["Hierarchical"],
         marker="*",
         label="Hierarchical")

plt.xlabel("Number of Clusters")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score Comparison")
plt.legend()
plt.savefig(
    "screenshots/Silhouette comparison (all algorithms)",
    dpi=300,
    bbox_inches="tight"
    )

plt.show()
```
